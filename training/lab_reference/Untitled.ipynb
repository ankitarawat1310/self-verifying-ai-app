```python
import torch, time
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

name = "Qwen/Qwen2.5-Coder-1.5B-Instruct"
tok = AutoTokenizer.from_pretrained(name)
bnb = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_quant_type="nf4",
                         bnb_4bit_compute_dtype=torch.bfloat16,
                         bnb_4bit_use_double_quant=True)
t0 = time.time()
model = AutoModelForCausalLM.from_pretrained(name, quantization_config=bnb,
                                             device_map={"": 0}, dtype=torch.bfloat16)
print("load seconds:", round(time.time() - t0, 1))

model = prepare_model_for_kbit_training(model)
model = get_peft_model(model, LoraConfig(r=16, lora_alpha=32, lora_dropout=0.05,
                                         target_modules="all-linear", task_type="CAUSAL_LM"))
model.print_trainable_parameters()

text = "def add(a, b):\n    return a + b\n\n" * 400
ids = tok(text, return_tensors="pt", truncation=True, max_length=1024).input_ids.cuda()
batch = ids.repeat(2, 1)  # batch of 2 sequences, 1024 tokens each
opt = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=1e-4)
model.train()
torch.cuda.reset_peak_memory_stats()
times = []
for step in range(12):
    t = time.time()
    loss = model(input_ids=batch, labels=batch).loss
    loss.backward(); opt.step(); opt.zero_grad()
    torch.cuda.synchronize(); times.append(time.time() - t)
    print(step, round(loss.item(), 3), round(times[-1], 2), "s")
avg = sum(times[2:]) / len(times[2:])
print("avg step s:", round(avg, 2), "| tokens/s:", round(batch.numel() / avg))
print("peak GPU mem GB:", round(torch.cuda.max_memory_allocated() / 1e9, 1))
```

```python
%%writefile /app/unsloth_30b_test.py
import unsloth  # must be imported first
from unsloth import FastModel
import sys, time, json, torch, psutil

name = sys.argv[1] if len(sys.argv) > 1 else "Qwen/Qwen3-Coder-30B-A3B-Instruct"
lengths = [1024, 2048]
out = {"model": name, "runs": []}

def ram():
    m = psutil.virtual_memory()
    return round(m.used/1e9, 1), round(m.available/1e9, 1)

print("start RAM used/avail GB:", ram(), flush=True)
t0 = time.time()
model, tok = FastModel.from_pretrained(
    model_name=name, max_seq_length=2048, load_in_4bit=True, full_finetuning=False)
out["load_seconds"] = round(time.time() - t0, 1)
out["gpu_after_load_gb"] = round(torch.cuda.memory_allocated() / 1e9, 1)
print("load seconds:", out["load_seconds"], "| GPU GB:", out["gpu_after_load_gb"],
      "| RAM used/avail:", ram(), flush=True)

model = FastModel.get_peft_model(
    model, r=16, lora_alpha=16, lora_dropout=0, bias="none",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
    use_gradient_checkpointing="unsloth", random_state=3407)
tk = getattr(tok, "tokenizer", tok)
text = "def add(a, b):\n    return a + b\n\n" * 2000
params = [p for p in model.parameters() if p.requires_grad]
opt = torch.optim.AdamW(params, lr=1e-4)
model.train()

for L in lengths:
    rec = {"seq_len": L}
    try:
        ids = tk(text, return_tensors="pt", truncation=True, max_length=L).input_ids.cuda()
        torch.cuda.reset_peak_memory_stats()
        times = []
        for s in range(6):
            t = time.time()
            loss = model(input_ids=ids, labels=ids).loss
            loss.backward(); opt.step(); opt.zero_grad()
            torch.cuda.synchronize(); times.append(time.time() - t)
            print(L, s, round(loss.item(), 3), round(times[-1], 2), "s", flush=True)
        avg = sum(times[2:]) / len(times[2:])
        rec.update(avg_step_s=round(avg, 2), tokens_per_s=round(ids.numel() / avg),
                   peak_gpu_gb=round(torch.cuda.max_memory_allocated() / 1e9, 1))
    except torch.cuda.OutOfMemoryError:
        rec["error"] = "OOM"; torch.cuda.empty_cache()
    except Exception as e:
        rec["error"] = repr(e)[:300]
    out["runs"].append(rec); print(rec, flush=True)
    json.dump(out, open("/app/results_" + name.replace("/", "_") + ".json", "w"), indent=1)
print("DONE", flush=True)
```

```python
%%writefile /app/unsloth_bench.py
import unsloth
from unsloth import FastModel
import sys, time, json, torch

name = sys.argv[1]
lengths = [int(x) for x in sys.argv[2].split(",")]
tag = name.replace("/", "_")
out = {"model": name, "runs": []}
def save(): json.dump(out, open(f"/app/bench_{tag}.json", "w"), indent=1)

t0 = time.time()
model, tok = FastModel.from_pretrained(model_name=name, max_seq_length=max(lengths),
                                       load_in_4bit=True, full_finetuning=False)
out["load_seconds"] = round(time.time() - t0, 1)
out["gpu_after_load_gb"] = round(torch.cuda.memory_allocated() / 1e9, 1)
print("load:", out["load_seconds"], "s | GPU GB:", out["gpu_after_load_gb"], flush=True)
model = FastModel.get_peft_model(model, r=16, lora_alpha=16, lora_dropout=0, bias="none",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
    use_gradient_checkpointing="unsloth", random_state=3407)
tk = getattr(tok, "tokenizer", tok)
text = "def add(a, b):\n    return a + b\n\n" * 8000
opt = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=1e-4)
model.train()
for L in lengths:
    rec = {"seq_len": L}
    try:
        ids = tk(text, return_tensors="pt", truncation=True, max_length=L).input_ids.cuda()
        torch.cuda.reset_peak_memory_stats(); times = []
        for s in range(6):
            t = time.time()
            loss = model(input_ids=ids, labels=ids).loss
            loss.backward(); opt.step(); opt.zero_grad()
            torch.cuda.synchronize(); times.append(time.time() - t)
            print(L, s, round(times[-1], 2), "s", flush=True)
        avg = sum(times[2:]) / len(times[2:])
        rec.update(avg_step_s=round(avg, 2), tokens_per_s=round(ids.numel() / avg),
                   peak_gpu_gb=round(torch.cuda.max_memory_allocated() / 1e9, 1))
        out["runs"].append(rec); print(rec, flush=True); save()
    except Exception as e:
        rec["error"] = "OOM" if "out of memory" in str(e).lower() else repr(e)[:300]
        out["runs"].append(rec); print(rec, flush=True); save(); break
try:
    model.eval(); FastModel.for_inference(model)
    p = tk("Write a Python function that reads a CSV file and returns a list of dicts.",
           return_tensors="pt").to("cuda")
    model.generate(**p, max_new_tokens=32, do_sample=False)  # warm-up
    t = time.time(); g = model.generate(**p, max_new_tokens=200, do_sample=False)
    n = g.shape[1] - p["input_ids"].shape[1]
    out["gen_tokens_per_s"] = round(n / (time.time() - t), 1)
except Exception as e:
    out["gen_error"] = repr(e)[:300]
print("gen:", out.get("gen_tokens_per_s"), out.get("gen_error"), flush=True)
save(); print("DONE", flush=True)
```

```python
%%writefile /app/inspect_data.py
import json
from datasets import load_dataset
from transformers import AutoTokenizer

ds = load_dataset("open-thoughts/OpenThoughts-Agent-SFT-10K", split="train")
print(ds); print(ds.column_names)
for k, v in ds[0].items():
    print("COL", k, type(v).__name__, str(v)[:200].replace("\n", " "))

col = next(c for c in ds.column_names
           if isinstance(ds[0][c], list) and ds[0][c] and isinstance(ds[0][c][0], dict))
print("conversation column:", col)
tok = AutoTokenizer.from_pretrained("Qwen/Qwen3-Coder-30B-A3B-Instruct")
lens, turns = [], []
for i in range(min(len(ds), 2000)):
    msgs = ds[i][col]
    text = "\n".join(str(m.get("content", m.get("value", ""))) for m in msgs)
    lens.append(len(tok(text).input_ids)); turns.append(len(msgs))
lens.sort()
pct = lambda p: lens[int(len(lens) * p)]
print("rows sampled:", len(lens), "| median:", pct(.5), "| p90:", pct(.9), "| max:", lens[-1])
for cap in (4096, 8192, 16384, 24576, 32768):
    print("fits under", cap, ":", round(100 * sum(l <= cap for l in lens) / len(lens), 1), "%")
print("avg turns:", round(sum(turns) / len(turns), 1))
json.dump({"lens": lens}, open("/app/dataset_stats.json", "w"))
```

```python
%%writefile /app/inspect2.py
import re, collections, random
from datasets import load_dataset
from transformers import AutoTokenizer
ds = load_dataset("open-thoughts/OpenThoughts-Agent-SFT-10K", split="train")
print("result values:", collections.Counter(str(x) for x in ds["result"]).most_common(5))
print("trace_source:", collections.Counter(ds["trace_source"]).most_common(8))
base = [re.sub(r"_copy\d+$", "", t) for t in ds["task"]]
print("unique tasks:", len(set(base)), "of", len(base))
print("task prefixes:", collections.Counter(b.split("-")[0] for b in base).most_common(8))
m = ds[0]["conversations"]
print("message keys:", list(m[0].keys()), "| roles:", [x.get("role") for x in m[:6]])
tok = AutoTokenizer.from_pretrained("Qwen/Qwen3-Coder-30B-A3B-Instruct")
idx = random.Random(0).sample(range(len(ds)), 2000)
lens = [len(tok("\n".join(str(x.get("content", "")) for x in ds[i]["conversations"])).input_ids) for i in idx]
s = sorted(lens); n = len(s)
print("RANDOM sample | median:", s[n//2], "| mean:", round(sum(s)/n), "| p90:", s[int(n*.9)], "| max:", s[-1])
for cap, sp in {12288: 2574, 16384: 2347, 24576: 1948}.items():
    keep = [l for l in lens if l <= cap]; trunc = [min(l, cap) for l in lens]; k = 10000 / n
    print(f"cap {cap}: DROP -> {round(100*len(keep)/n,1)}% = {round(len(keep)*k)} rows, {round(sum(keep)*k/1e6)}M tokens, ~{round(sum(keep)*k/(sp*3600),1)} h | TRUNCATE -> 10000 rows, {round(sum(trunc)*k/1e6)}M tokens, ~{round(sum(trunc)*k/(sp*3600),1)} h")
```
